### 1. Create a program that gives different messages based on time of day

```python
from datetime import datetime

# Get current hour
now = datetime.now()
hour = now.hour

print("Current Time:", now.strftime("%I:%M %p"))
# Decide message based on time
if 5 <= hour < 12:
    print("Good Morning! Have a wonderful day ahead ☀️")
elif 12 <= hour < 16:
    print("Good Afternoon! Hope your day is going well 🌼")
elif 16 <= hour < 20:
    print("Good Evening! Relax and unwind ✨")
else:
    print("Good Night! Time to rest and recharge 🌙")
```

### 2. Build a number guessing game using if-else statements

```python
import random

secret = random.randint(1,100)

while True:
    guess = int(input("Guess a number:"))
    if guess == secret:
        print("Correct! You guessed it right!")
        break
    elif guess < secret:
        print("Guess Higher!")
    else:
        print("Guess Lower!")
```

### 4. Create a function that calculates area of rectangle

```python
l=5
b=2
def area_rectangle(l,b):
    if l <= 0 or b <= 0:
        return "Invalid dimensions! Length and breadth must be positive."
    return l*b

print(f"The area of rectangle with length {l} & breadth {b} is {area_rectangle(l,b)}.")
```
