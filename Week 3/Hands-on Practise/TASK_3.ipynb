### 2. Download a simple dataset (like student marks or sales data)

```python
## Downloaded Amazon Store Sales Data
import numpy as np
import pandas as pd
```

### 3. Load the dataset and display first few rows

```python
# Loading the dataset

data = pd.read_csv("Amazon Store Sales Data.csv")

print("---------------------------- THE FIRST 5 RECORDS OF DATA ----------------------------")
print(data.head(5))
```

### 4. Check basic information: how many rows, columns, data types

```python
print("\n ---------------- Dataset Shape (Rows, Columns):-----------------")
print(data.shape)

print("\n------------------------- Column Names: -------------------------")
print(data.columns)

print("\n------------------------- Data Types: -------------------------")
print(data.dtypes)

print("\n------------------------- Full Info:-------------------------")
data.info()
```

### 5. Find and handle missing values in the dataset

```python
print("\n -------Missing Values in Each Column:---------")
print(data.isnull().sum())
```

### 6. Calculate average, highest, lowest values for numerical columns

```python
print("\n --------------- Summary Statistics for Numerical Columns: ----------------")
print(data.describe())

# Manually calculate:
print("\n----------- Averages: -----------")
print(data.mean(numeric_only=True))

print("\n-------- Maximum values: --------")
print(data.max(numeric_only=True))

print("\n--------- Minimum values:---------")
print(data.min(numeric_only=True))
```
